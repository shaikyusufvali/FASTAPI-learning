# Day 04: Query Parameters in FastAPI

## What are Query Parameters?

Query Parameters are values passed to an API through the URL.

They start after the question mark (?).

Example:
http://127.0.0.1:8000/users?name=Yusuf&age=23

Here:
- /users = API Endpoint
- ? = Query Parameter Starting Point
- name = Parameter Name
- Yusuf = Parameter Value
- age = Parameter Name
- 23 = Parameter Value

### Purpose:
1. Pass additional information.
2. Search specific records.
3. Customize API requests.
4. Make API endpoints flexible.

```python
# Parameter API
from fastapi import FastAPI
app = FastAPI()
@app.get("/products")
def get_products(Category: str):
    return {
        "Category": Category,
        "message": "Products retrieved successfully"
    }
```

## Multiple Query Parameters

FastAPI allows multiple query parameters
in a single API request.

Example:

/student?name=Yusuf&age=23&course=AI

We can receive multiple values using
function parameters.

```python
@app.get("/student")
def get_student(
    name: str,
    age: int,
    course: str
):
    return {
        "name": name,
        "age": age,
        "course": course
    }
```

## Required Query Parameters

Required parameters must be provided by the client.

Example:

name: str

If the client does not provide name,
FastAPI returns HTTP 422 Validation Error.

```python
@app.get("/required")
def required_parameter(name: str):
    return {
        "name": name
    }
```

## Optional Query Parameters

Optional parameters are not compulsory.

We use None as the default value.

Syntax:

parameter: type | None = None

If the client does not provide a value,
the default value will be None.

```python
@app.get("/optional")
def optional_parameter(
    name: str,
    age: int | None = None
):
    return {
        "name": name,
        "age": age
    }
```

## Default Values

Default values are automatically used
when the client does not provide a parameter.

Example:

page: int = 1
limit: int = 10

If the client does not provide page and limit,
FastAPI uses 1 and 10.

```python
@app.get("/pagination")
def get_data(
    page: int = 1,
    limit: int = 10
):
    return {
        "page": page,
        "limit": limit
    }
```

## Query Parameter Validation

FastAPI provides Query() for parameter validation.

We can validate:
- Minimum value
- Maximum value
- Minimum string length
- Maximum string length

Invalid inputs generate HTTP 422.

```python
from fastapi import Query

@app.get("/validate")
def validate_user(
    age: int = Query(ge=18, le=60),
    name: str = Query(min_length=3, max_length=20)
):
    return {
        "name": name,
        "age": age,
        "message": "Validation Successful"
    }
```
