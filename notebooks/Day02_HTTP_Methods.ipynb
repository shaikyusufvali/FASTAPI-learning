```python
from fastapi import FastAPI

app = FastAPI()

print("FastAPI Application Created Successfully!")
```

```python
users = [
    {"id": 1, "name": "Yusuf", "age": 23},
    {"id": 2, "name": "Rahul", "age": 24}
]

print(users)
```

```python
@app.get("/users")
def get_users():
    return users
```

```python
# (POST Method)
@app.post("/users")
def create_user(user: dict):
    users.append(user)
    return {"message": "User created successfully", "user": user}
```

```python
# put method
@app.put("/users/{user_id}")
def update_user(user_id: int, updated_user: dict):

    for user in users:

        if user["id"] == user_id:

            user.update(updated_user)

            return {
                "message": "User updated successfully",
                "user": user
            }

    return {"message": "User not found"}
```

```python
# (PATCH Method)
@app.patch("/users/{user_id}")
def partial_update(user_id: int, data: dict):

    for user in users:

        if user["id"] == user_id:

            user.update(data)

            return {
                "message": "Partial update successful",
                "user": user
            }

    return {"message": "User not found"}
```

```python
# Delete Method
@app.delete("/users/{user_id}")
def delete_user(user_id: int):
    for user in users:

        if user["id"] == user_id:

            users.remove(user)

            return {"message": "User deleted successfully"}

    return {"message": "User not found"}
```

# Day 02 - Important Concepts

### Request
Frontend sends data to backend.

### Response
Backend sends result to frontend.

### Endpoint
API URL used to access a particular operation.

### Path Parameter
Dynamic value passed through URL.

Example: /users/3

### Request Body
Data sent by client to backend.

### Dictionary
Python data structure containing key-value pairs.

### append()
Adds new item to a list.

### update()
Updates dictionary values.

### remove()
Removes an item from a list.

### Important Difference

GET = Read
POST = Create
PUT = Complete Update
PATCH = Partial Update
DELETE = Remove
