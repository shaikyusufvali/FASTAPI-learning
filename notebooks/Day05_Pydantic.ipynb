# Day 05: Pydantic & Data Validation

## What is Pydantic?

Pydantic is a Python library used for data validation
and data parsing.

It checks whether incoming data matches the
expected data types and constraints.

### Why do we use Pydantic?

1. Data validation
2. Type checking
3. Required field checking
4. Default values
5. Data structure validation
6. API request body validation

### Real-Time Example

A user sends:

{
    "name": "Yusuf",
    "age": 23,
    "email": "yusuf@gmail.com"
}

Pydantic checks whether the received data
matches the expected structure.

## What is BaseModel?

BaseModel is the main class provided by Pydantic.

We create data models by inheriting from BaseModel.

These models define:
- Field names
- Data types
- Required fields
- Optional fields
- Default values

Syntax:

class Student(BaseModel):
    name: str
    age: int

```python
from pydantic import BaseModel
class Student(BaseModel):
    name: str
    age: int
```

```python
# creating pydantic model for student
student = Student(name="Yusuf", age=23)
print(Student)
print(student.name)
print(student.age)
```

## Data Validation

Data validation means checking whether
the received data is valid.

Pydantic validates data using
the types and constraints defined in the model.

Example:

name: str
age: int

Expected:
name = String
age = Integer

Invalid data produces a ValidationError.

```python
from pydantic import BaseModel, ValidationError

class Student(BaseModel):
    name: str
    age: int


try:
    student = Student(
        name="Yusuf",
        age="twenty"
    )

    print(student)

except ValidationError as error:
    print(error)
```

## Required Fields

Required fields are mandatory fields.

If we define a field without a default value,
Pydantic considers it required.

Example:

class Student(BaseModel):
    name: str
    age: int

Both name and age are required.

```python
class Student(BaseModel):
    name: str
    age: int | None = None

student = Student(name="Yusuf")

print(student)
```

## Optional Fields

Optional fields are fields that do not require
a value from the user.

We can use:

field: type | None = None

Example:

email: str | None = None

The user can provide an email or leave it empty.

```python
class Student(BaseModel):
    name: str
    age: int
    email: str | None = None


student = Student(
    name="Yusuf",
    age=23
)

print(student)
```

## Default Values

Default values are automatically assigned
when the user does not provide a value.

Example:

age: int = 18

If age is missing, Pydantic uses 18.

```python
class Student(BaseModel):
    name: str
    age: int = 18


student = Student(name="Yusuf")

print(student)
```

## Field Validation

Pydantic provides Field() to define
constraints on model fields.

We can specify:

1. Minimum value
2. Maximum value
3. Minimum string length
4. Maximum string length
5. Field descriptions

```python
from pydantic import BaseModel, Field

class Student(BaseModel):

    name: str = Field(
        min_length=3,
        max_length=20
    )

    age: int = Field(
        ge=18,
        le=60
    )


student = Student(
    name="Yusuf",
    age=23
)

print(student)
```

## Pydantic with FastAPI

FastAPI uses Pydantic models to validate
incoming request bodies.

When a client sends JSON data:

1. FastAPI receives the HTTP request.
2. Pydantic validates the request body.
3. Valid data is passed to the API function.
4. The function processes the data.
5. FastAPI returns a JSON response.

We commonly use Pydantic models with POST APIs.

```python
from fastapi import FastAPI
from pydantic import BaseModel, Field

app = FastAPI()


class Student(BaseModel):
    name: str
    age: int = Field(ge=18, le=60)
    course: str


@app.post("/students")
def create_student(student: Student):

    return {
        "message": "Student created successfully",
        "student_name": student.name,
        "student_age": student.age,
        "student_course": student.course
    }
```

## Nested Pydantic Models

Nested models mean using one Pydantic model
inside another Pydantic model.

They are useful when JSON data contains
multiple levels of objects.

Example:

Student
    |
    |-- name
    |-- age
    |
    |-- address
            |
            |-- city
            |-- pincode

```python
from pydantic import BaseModel

class Adress(BaseModel):
    state: str
    pin_code: int

class Student(BaseModel):
    name: str
    age: int
    address: Adress
Student = Student(
    name="Yusuf",
    age=23,
    address={
        "state": "Lagos",
        "pin_code": 100001
    }
)
print(Student.model_dump())
```

## Model Serialization

Serialization means converting a Pydantic model
into a Python dictionary.

Pydantic V2 provides:

model.model_dump()

This is useful when sending data to databases,
other services, or JSON responses.

```python
from pydantic import BaseModel

class Address(BaseModel):
    city: str
    pincode: int

class Student(BaseModel):
    name: str
    age: int
    address: Address
```

```python
student = Student(
    name="Yusuf",
    age=23,
    address={
        "city": "Hyderabad",
        "pincode": 500001
    }
)

print(student)
```

```python
student_data = student.model_dump()

print(student_data)
print(type(student_data))
```
